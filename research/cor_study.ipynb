```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_selection import mutual_info_regression, mutual_info_classif
```

```python
DATA_PERIOD = 10
_download = {"auto_adjust": True, "progress": False, "period": f"{DATA_PERIOD}y"}
vix_data = yf.download("^VIX", **_download)
spx_data = yf.download("^GSPC", **_download)
skew_data = yf.download("^SKEW", **_download)
vix9d_data = yf.download("^VIX9D", **_download)
vvix_data = yf.download("^VVIX", **_download)
SDEX_data = yf.download("^SDEX", **_download)
```

```python
df = pd.concat(
    {
        "VIX": vix_data["Close"].iloc[:, 0],
        "SPX": spx_data["Close"].iloc[:, 0],
        "SKEW": skew_data["Close"].iloc[:, 0],
        "VIX9D": vix9d_data["Close"].iloc[:, 0],
        "VVIX": vvix_data["Close"].iloc[:, 0],
        "SDEX": SDEX_data["Close"].iloc[:, 0],
    },
    axis=1,
)
df["SPX_ret"] = np.log(df["SPX"]/df["SPX"].shift(1))
df["RV"] = df["SPX_ret"].rolling(22).std()
df["VRP"] = df["VIX"] - df["RV"]
df["lagged_VRP"] = df["VIX"].shift(22) - df["RV"]
df["vvix_vix"] = df["VIX"]/df["VVIX"]
df["skew_sdex"] = df["SKEW"]/df["SDEX"]


df["Label_ret"] = df["SPX"].shift(-22) - df["SPX"]


labels = ["Label_ret"]
features = df.columns.drop(labels)
# print(df.iloc[200:210, :])
```
